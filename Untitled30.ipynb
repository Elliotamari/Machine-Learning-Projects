```python
# =============================================================================
# 0. Configuration & Imports
# =============================================================================
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split, StratifiedShuffleSplit, RandomizedSearchCV
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, mean_absolute_percentage_error
import json

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 180)
DATA_PATH = Path('/content/drive/MyDrive/home_insurance.csv')
PICKLE_PATH = "/content/drive/MyDrive/final_df.pkl"
CSV_PATH    = "/content/drive/MyDrive/final_df.csv"
```

```python
# =============================================================================
# 1. Load & Initial Inspection
# =============================================================================
df = pd.read_csv(DATA_PATH, low_memory=False)
print("Raw shape:", df.shape)
```

```python
# A) Column Summary Table: dtype, missing count, missing %, unique count
# ----------------------------------------------------------------------------
n_rows = df.shape[0]

# Build a summary DataFrame
col_summary = pd.DataFrame({
    "dtype": df.dtypes,
    "missing_count": df.isna().sum(),
    "missing_pct": df.isna().sum() / n_rows * 100,
    "unique_count": df.nunique(dropna=False)
})

# Sort by missing_pct descending
col_summary = col_summary.sort_values(by="missing_pct", ascending=False)
pd.set_option('display.max_rows', None, 'display.width', 120)
print(col_summary)
```

```python
# B) Value Counts for Categorical (or low‐cardinality) Columns
# ----------------------------------------------------------------------------
to_inspect = [
    col for col in df.columns
    if (df[col].dtype == "object") or (df[col].nunique() <= 10)
]

for col in to_inspect:
    print(f"\n--- VALUE COUNTS for column: {col} (dtype={df[col].dtype}, uniques={df[col].nunique()}) ---")
    print(df[col].value_counts(dropna=False).head(20))  # top 20 categories, including NaN
```

```python
# =============================================================================
# 3. Data Cleaning & Preprocessing (Initial Drops & Parsing Dates)
# =============================================================================

#  Drop columns with extreme missingness or unwanted features
#  (IDs, leakage columns, high‐cardinality/sparse categoricals, raw date‐strings)
# ----------------------------------------------------------------------------
drop_initial = [
    "i", "Police", "QUOTE_DATE", "MTA_DATE", "POL_STATUS", "P1_POLICY_REFUSED",
    "OCC_STATUS", "PROP_TYPE", "ROOF_CONSTRUCTION", "WALL_CONSTRUCTION",
    "CAMPAIGN_DESC", "P1_PT_EMP_STATUS", "CLERICAL", "PAYMENT_FREQUENCY",
    "MTA_FAP", "MTA_APRP", "RISK_RATED_AREA_B", "RISK_RATED_AREA_C"
]
df.drop(columns=drop_initial, inplace=True, errors="ignore")

# Drop rows with >26% missing in key cols
# ----------------------------------------------------------------------------
key_nulls = df.isna().sum().value_counts().idxmax()
df.dropna(subset=df.columns[df.isna().sum() == key_nulls], inplace=True)

# Parse dates & derive ages
# ----------------------------------------------------------------------------
df["COVER_START"] = pd.to_datetime(df["COVER_START"], format="%d/%m/%Y", errors="coerce")
df["P1_DOB"]      = pd.to_datetime(df["P1_DOB"],      format="%d/%m/%Y", errors="coerce")
df["YEARBUILT"]   = pd.to_numeric(df["YEARBUILT"],    errors="coerce")

df["BUILDING_AGE"] = df["COVER_START"].dt.year - df["YEARBUILT"]
df["AGE_AT_QUOTE"] = df["COVER_START"].dt.year - df["P1_DOB"].dt.year

df.drop(columns=["COVER_START", "P1_DOB", "YEARBUILT"], inplace=True, errors="ignore")
print("Shape after date parsing:", df.shape)
```

```python
# =============================================================================
# 4. Feature Engineering
# =============================================================================

# -- Count total active add-ons (and binarise each Y/N column)
# ----------------------------------------------------------------------------
addon_cols = [
    "LEGAL_ADDON_PRE_REN", "LEGAL_ADDON_POST_REN",
    "HOME_EM_ADDON_PRE_REN", "HOME_EM_ADDON_POST_REN",
    "GARDEN_ADDON_PRE_REN", "GARDEN_ADDON_POST_REN",
    "KEYCARE_ADDON_PRE_REN", "KEYCARE_ADDON_POST_REN",
    "HP1_ADDON_PRE_REN", "HP1_ADDON_POST_REN",
    "HP2_ADDON_PRE_REN", "HP2_ADDON_POST_REN",
    "HP3_ADDON_PRE_REN", "HP3_ADDON_POST_REN"
]

# Sum up booleans where the cell == "Y"
df["TOTAL_ADDONS_SELECTED"] = df[addon_cols].eq("Y").sum(axis=1)

# Convert each column to 1/0
for col in addon_cols:
    df[col] = df[col].map({"Y": 1, "N": 0}).fillna(0).astype(int)


# -- Map & binarise other flags
# ----------------------------------------------------------------------------
df["POLICYHOLDER_GENDER"] = df["P1_SEX"].map({"M":1,"F":0}).fillna(-1).astype(int)
df["EMPLOYMENT_STATUS"]   = df["P1_EMP_STATUS"].map({"R":"RETIRED","E":"EMPLOYED","S":"STUDENT"}).fillna("OTHER")
df["MARITAL_STATUS"]       = df["P1_MAR_STATUS"].replace(
    {"P":"PARTNER","O":"PARTNER","W":"PREVIOUSLY_MARRIED","D":"PREVIOUSLY_MARRIED","S":"PREVIOUSLY_MARRIED"}
).fillna("OTHER")
df["PAYMENT.METHOD"]       = df["PAYMENT_METHOD"].map({"PureDD":"PURE_DD","NonDD":"NON_DD","DD-Other":"DD_OTHER"}).fillna("UNKNOWN")
top2 = df["OWNERSHIP_TYPE"].value_counts().nlargest(2).index
df["OWNERSHIP.TYPE"]       = df["OWNERSHIP_TYPE"].where(df["OWNERSHIP_TYPE"].isin(top2),"OTHER").fillna("OTHER")
df["IS_BUSINESS_USE"]      = df["BUS_USE"].map({"Y":1}).fillna(0).astype(int)
df["HAS_FLOODING_RISK"]    = df["FLOODING"].map({"Y":1}).fillna(0).astype(int)
df["HAS_SUBSIDENCE_RISK"]  = df["SUBSIDENCE"].map({"Y":1}).fillna(0).astype(int)
df["CLAIMED_PAST_3YRS"]    = df["CLAIM3YEARS"].map({"Y":1,"N":0}).fillna(0).astype(int)
df["HAS_MTA_CHANGE"]       = df["MTA_FLAG"].map({"Y":1,"N":0}).fillna(0).astype(int)
for c in ["BUILDINGS_COVER","CONTENTS_COVER","AD_BUILDINGS","AD_CONTENTS"]:
    df[f"HAS_{c}"] = df[c].map({"Y":1,"N":0}).fillna(0).astype(int)
df["HAS_PAYING_GUESTS"]    = df["PAYING_GUESTS"].fillna(0).astype(int)

alarm_flag = df["APPR_ALARM"].eq("Y").fillna(False)
lock_flag  = df["APPR_LOCKS"].eq("Y").fillna(False)

df["HAS_SECURITY_FEATURES"] = (alarm_flag | lock_flag).astype(int)

df["IS_FULLY_INSURED_BUILDING"] = (df["SUM_INSURED_BUILDINGS"]==1_000_000).astype(int)
df["PARTIAL_BUILDING_INSURED_AMOUNT"] = df["SUM_INSURED_BUILDINGS"].where(df["SUM_INSURED_BUILDINGS"]<1_000_000,0)
df["HAS_STANDARD_CONTENTS"] = (df["SUM_INSURED_CONTENTS"]==50_000).astype(int)
df["HAS_HIGH_CONTENTS"]     = (df["SUM_INSURED_CONTENTS"]==100_000).astype(int)
df["HAS_NO_CONTENTS"]       = (df["SUM_INSURED_CONTENTS"]==0).astype(int)
df = df[df["MAX_DAYS_UNOCC"]>=0]
df["UNOCCUPANCY.TYPE"]      = pd.cut(df["MAX_DAYS_UNOCC"],
                                     bins=[-1,0,60,df["MAX_DAYS_UNOCC"].max()],
                                     labels=["NEVER_VACANT","SHORT_VACANT","LONG_VACANT"])

# -- Drop raw columns now engineered
# ----------------------------------------------------------------------------
drop_after = [
    "P1_SEX","P1_EMP_STATUS","P1_MAR_STATUS","PAYMENT_METHOD","OWNERSHIP_TYPE",
    "BUS_USE","FLOODING","SUBSIDENCE","CLAIM3YEARS","MTA_FLAG",
    "BUILDINGS_COVER","CONTENTS_COVER","AD_BUILDINGS","AD_CONTENTS",
    "PAYING_GUESTS","APPR_ALARM","APPR_LOCKS","SUM_INSURED_BUILDINGS",
    "SUM_INSURED_CONTENTS","MAX_DAYS_UNOCC","LISTED","SAFE_INSTALLED",
    "SEC_DISC_REQ","NEIGH_WATCH"
]
df.drop(columns=drop_after, inplace=True, errors="ignore")
print("Shape after feature engineering:", df.shape)
```

```python
# -- One-hot encode
# ----------------------------------------------------------------------------
multi_cat = ["EMPLOYMENT_STATUS","MARITAL_STATUS","PAYMENT.METHOD","OWNERSHIP.TYPE","UNOCCUPANCY.TYPE"]
df = pd.get_dummies(df, columns=multi_cat, drop_first=True)
df[df.select_dtypes("bool").columns] = df.select_dtypes("bool").astype(int)
print("Shape after encoding:", df.shape)
```

```python
# =============================================================================
# 5. Outliers & Target Transform
# =============================================================================
df = df[df["LAST_ANN_PREM_GROSS"]>0].copy()
df["LOG_SPEC_SUM_INSURED"] = np.log1p(df.pop("SPEC_SUM_INSURED"))
df["LOG_SPEC_ITEM_PREM"]   = np.log1p(df.pop("SPEC_ITEM_PREM"))
df["LOG_UNSPEC_HRP_PREM"]  = np.log1p(df.pop("UNSPEC_HRP_PREM"))

cap = df["LAST_ANN_PREM_GROSS"].quantile(0.99)
df["PREMIUM"] = df["LAST_ANN_PREM_GROSS"].clip(upper=cap)

# Drop now-redundant raw target
df.drop(columns=["LAST_ANN_PREM_GROSS"], inplace=True)
```

```python
# =============================================================================
# 6. Correlation Prune & Zero-Variance
# =============================================================================
corr = df.corr().abs()
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
to_drop_corr = [col for col in upper.columns if any(upper[col]>0.90)]
df.drop(columns=to_drop_corr, inplace=True, errors="ignore")

const = df.columns[df.nunique()==1]
df.drop(columns=const, inplace=True, errors="ignore")
print("Shape after pruning:", df.shape)
```

```python
# =============================================================================
# 7. Save Cleaned Data
# =============================================================================
df.to_pickle(PICKLE_PATH)
df.to_csv(CSV_PATH, index=False)
print("✅ Data saved.")
```

# **MODELLING**

```python
# =============================================================================
# 8. Subsampling & Model Evaluation
# =============================================================================
def stratified_subsample(data, frac=0.1, n_bins=10, random_state=42):
    if frac >= 1.0:
        return data.copy()
    tmp = data.copy()
    tmp["prem_bin"] = pd.qcut(tmp["PREMIUM"], q=n_bins, labels=False, duplicates="drop")
    splitter = StratifiedShuffleSplit(n_splits=1, test_size=frac, random_state=random_state)
    idx = next(splitter.split(tmp, tmp["prem_bin"]))[1]
    return tmp.iloc[idx].drop(columns=["prem_bin"])

def evaluate_model(name, model, data, fractions=[0.1,0.25,0.5,0.75,1.0], random_state=42):
    results = []
    for f in fractions:
        sub = stratified_subsample(data, frac=f, random_state=random_state)
        X = sub.drop(columns=["PREMIUM"])
        y = sub["PREMIUM"]
        X_tr,X_te,y_tr,y_te = train_test_split(X,y,test_size=0.2,random_state=random_state)
        m = model.__class__(**model.get_params())
        if hasattr(m, "random_state"):
            setattr(m, "random_state", random_state)
        m.fit(X_tr, y_tr)
        p = m.predict(X_te)
        results.append({
            "Model": name, "Fraction": f,
            "R2": r2_score(y_te,p),
            "MAE": mean_absolute_error(y_te,p),
            "RMSE": np.sqrt(mean_squared_error(y_te,p)),
            "MAPE": mean_absolute_percentage_error(y_te,p)
        })
    return pd.DataFrame(results)
```

```python
models = [
    ("LinearRegression", LinearRegression()),
    ("Lasso",            Lasso(alpha=1.0)),
    ("RandomForest",     RandomForestRegressor(n_estimators=100, random_state=42)),
    ("XGBoost",          XGBRegressor(n_estimators=100, random_state=42)),
    ("KNN",              KNeighborsRegressor()),
    ("SVR",              SVR(kernel='rbf')),
]

all_results = pd.concat(
    [evaluate_model(n, m, df) for n, m in models],
    ignore_index=True
)

# Print per-model
for name in all_results["Model"].unique():
    print(f"\n=== Results for {name} ===")
    print(all_results[all_results["Model"]==name]
          .to_string(index=False, float_format="%.4f"))
```

# **HYPERPARAMETER TUNING**

```python
# =============================================================================
# XGBoost Hyperparameter Tuning with RandomizedSearchCV (100% data)
# =============================================================================
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from xgboost import XGBRegressor
import numpy as np

# 1) Define the hyperparameter distributions
param_dist_xgb = {
    'n_estimators':      [1000, 5000, 10000, 25000, 40000],
    'max_depth':         [3, 6, 10, 15],
    'learning_rate':     [0.005, 0.01, 0.05, 0.1],
    'subsample':         [0.6, 0.8, 1.0],
    'colsample_bytree':  [0.6, 0.8, 1.0]
}

# 2) Build your 100% subsample
full_df = stratified_subsample(df, frac=1.0)
X_full = full_df.drop(columns=["PREMIUM"])
y_full = full_df["PREMIUM"]

# 3) Hold out a test set
X_tr, X_te, y_tr, y_te = train_test_split(
    X_full, y_full, test_size=0.2, random_state=42
)

# 4) Configure RandomizedSearchCV
xgb = XGBRegressor(random_state=42, verbosity=0)
rand_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist_xgb,
    n_iter=72,                            # ~10% of the 720 combos
    scoring='r2',
    cv=3,
    random_state=42,
    n_jobs=1,
    verbose=2
)

# 5) Run the search
print("🔍 Tuning XGBoost on 100% of data …")
rand_search.fit(X_tr, y_tr)

# 6) Best CV results
best_params = rand_search.best_params_
best_cv_r2  = rand_search.best_score_
print(f"\n Best hyperparameters: {best_params}")
print(f"↳ CV R² (avg): {best_cv_r2:.4f}")

# 7) Evaluate on held-out test set
best_model = rand_search.best_estimator_
y_pred     = best_model.predict(X_te)

test_r2   = r2_score(y_te, y_pred)
test_mae  = mean_absolute_error(y_te, y_pred)
test_rmse = np.sqrt(mean_squared_error(y_te, y_pred))
test_mape = mean_absolute_percentage_error(y_te, y_pred)

print("\n📊 Test set performance:")
print(f" • R²   = {test_r2:.4f}")
print(f" • MAE  = ${test_mae:.2f}")
print(f" • RMSE = ${test_rmse:.2f}")
print(f" • MAPE = {test_mape:.2%}")
```

```python
import json

# 7) Save tuning summary & best params
tuning_summary = {
    "Fraction": 1.0,
    "CV_R2":    best_cv_r2,
    "Test_R2":  test_r2,
    "Test_MAE": test_mae,
    "Test_RMSE": test_rmse,
    "Test_MAPE": test_mape,
    **best_params
}

pd.DataFrame([tuning_summary]) \
  .to_csv("/content/drive/MyDrive/xgb_tuning_full_r2.csv", index=False)

with open("/content/drive/MyDrive/xgb_best_params_full_r2.json", "w") as f:
    json.dump(best_params, f, indent=2)

print("✅ Saved tuning summary and best params to Drive")
```
