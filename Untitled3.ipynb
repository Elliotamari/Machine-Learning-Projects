# **2. Data Cleaning and Preprocessing**

```python
import pandas as pd
import numpy as np

data = df.copy()

# 1) Parse timestamp & keep rows with valid target and time
data["Start_Time"] = pd.to_datetime(data["Start_Time"], errors="coerce")
data = data.dropna(subset=["Severity", "Start_Time"]).reset_index(drop=True)

# 2) Time-derived, leakage-safe features
data["Hour"]       = data["Start_Time"].dt.hour
data["Weekday"]    = data["Start_Time"].dt.weekday  # 0=Mon
data["Month"]      = data["Start_Time"].dt.month
data["Is_Weekend"] = data["Weekday"].isin([5, 6]).astype(int)

# 3) Remove known leakage / post-event fields
leak_cols = [
    "End_Time", "End_Lat", "End_Lng", "Distance(mi)", "Weather_Timestamp",
    "Civil_Twilight", "Nautical_Twilight", "Astronomical_Twilight"
]
for c in leak_cols:
    if c in data.columns:
        data.drop(columns=c, inplace=True, errors="ignore")

# 4) Remove non-predictive IDs / heavy free text (optional NLP later)
for c in ["ID", "Source", "Description"]:
    if c in data.columns:
        data.drop(columns=c, inplace=True, errors="ignore")

# 5) Normalize boolean road-context flags to 0/1
bool_cols = ["Amenity","Bump","Crossing","Give_Way","Junction","No_Exit","Railway",
             "Roundabout","Station","Stop","Traffic_Calming","Traffic_Signal","Turning_Loop"]
for c in bool_cols:
    if c in data.columns:
        data[c] = data[c].astype("boolean").astype("Int64").fillna(0).astype(int)

# 6) Quick diagnostics
def diag_report(frame: pd.DataFrame, target_col="Severity", top_n=15):
    print("Cleaning complete.")
    print("Shape:", frame.shape)
    if target_col in frame.columns:
        vc = frame[target_col].value_counts().sort_index()
        print("\nSeverity counts:")
        print(vc)
    print("\nTop missing values:")
    print(frame.isna().sum().sort_values(ascending=False).head(top_n))
    mem_mb = frame.memory_usage(deep=True).sum() / (1024**2)
    print(f"\nApprox memory usage: {mem_mb:.2f} MB")
    print("\nColumns (first 20):", list(frame.columns[:20]))

diag_report(data)

# Keep references for next steps
TARGET = "Severity"
```

```python
# =========================================================
#— Hard Drop of All Missing Rows
# =========================================================
# Prof note:
# This step *removes every row with ANY NaN*

import pandas as pd
import numpy as np

# Start from the already-cleaned 'data' produced in Step 2
hard = data.copy()

before_rows = len(hard)
rows_with_any_na = int(hard.isna().any(axis=1).sum())
cols_with_na = hard.columns[hard.isna().any()].tolist()

print("Columns that contain missing values BEFORE drop:", len(cols_with_na))
print(sorted(cols_with_na))

# Drop all rows with any NaN
hard = hard.dropna().reset_index(drop=True)

after_rows = len(hard)
dropped = before_rows - after_rows
pct_dropped = 100.0 * dropped / before_rows if before_rows else 0.0

print(f"Rows before: {before_rows:,}")
print(f"Rows with any NaN (would be dropped): {rows_with_any_na:,}")
print(f"Rows after : {after_rows:,}  (dropped {dropped:,} = {pct_dropped:.2f}%)")
print("Shape after:", hard.shape)

# Severity distribution before vs after
def dist(s):
    vc = s.value_counts().sort_index()
    pct = (vc / vc.sum() * 100).round(2)
    return pd.DataFrame({"count": vc, "percent": pct})

print("\nSeverity distribution BEFORE:")
display(dist(data['Severity']))

print("Severity distribution AFTER:")
display(dist(hard['Severity']))

# Sanity: there should be zero missing now
top_missing_after = hard.isna().sum().sort_values(ascending=False).head(10)
print("\nTop missing values AFTER (should be 0):")
print(top_missing_after)

mem_mb = hard.memory_usage(deep=True).sum() / (1024**2)
print(f"\nApprox memory usage AFTER: {mem_mb:.2f} MB")

# Replace the working dataset ONLY if you accept the trade-off
data = hard
```

# **3. Feature Engineering & Encoding**

```python
import pandas as pd
import numpy as np
from collections import OrderedDict

work = data.copy()

# A) Drop raw timestamp (already engineered), and high-card/low-signal columns
drop_cols = ["Start_Time", "Street", "Zipcode", "Country"]
work = work.drop(columns=[c for c in drop_cols if c in work.columns])

# B) Define target and candidate features
TARGET = "Severity"
y = work[TARGET].astype(int)
X = work.drop(columns=[TARGET])

# C) Numeric features (present-check)
num_cols = [
    "Start_Lat","Start_Lng",
    "Temperature(F)","Wind_Chill(F)","Humidity(%)","Pressure(in)",
    "Visibility(mi)","Wind_Speed(mph)","Precipitation(in)",
    "Hour","Weekday","Month","Is_Weekend"
]
num_cols = [c for c in num_cols if c in X.columns]

# D) Boolean road-context flags (already converted to 0/1; treat as numeric to avoid OHE blow-up)
bool_cols = [c for c in [
    "Amenity","Bump","Crossing","Give_Way","Junction","No_Exit","Railway",
    "Roundabout","Station","Stop","Traffic_Calming","Traffic_Signal","Turning_Loop"
] if c in X.columns]

# E) Categorical features (string-like)
cat_cols = [c for c in X.columns if c not in (num_cols + bool_cols)]

# F) Quick cardinality & missing diagnostics for categoricals
def cardinality_report(df, cols, top=15):
    card = []
    for c in cols:
        nuniq = df[c].nunique(dropna=True)
        nmiss = df[c].isna().sum()
        card.append((c, nuniq, nmiss))
    rep = pd.DataFrame(card, columns=["column","n_unique","n_missing"]).sort_values("n_unique", ascending=False)
    return rep

# Ensure categoricals are strings for OHE later
X_cat = X[cat_cols].copy()
for c in X_cat.columns:
    X_cat[c] = X_cat[c].astype(str)

cat_card = cardinality_report(X_cat, cat_cols)

print(" Feature sets prepared (not fitted yet).")
print(f"- Numeric cols      : {len(num_cols)}")
print(f"- Boolean-as-numeric: {len(bool_cols)}")
print(f"- Categorical cols  : {len(cat_cols)}")

print("\nTop 15 categorical cardinalities:")
display(cat_card.head(15))

print("\nSample rows (sanity check):")
display(X.head(3)[num_cols[:5] + bool_cols[:5] + cat_cols[:5]])

# Store for next step
FEATURE_SETS = {
    "numeric": num_cols + bool_cols,   # treat bools as numeric
    "categorical": cat_cols
}
```

```python

import numpy as np, pandas as pd, sklearn
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Use the feature sets you already computed
numeric_features      = FEATURE_SETS["numeric"]
categorical_features  = FEATURE_SETS["categorical"]

# Split (stratified) BEFORE fitting the preprocessor (prevents leakage)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

def dist(s):
    vc = s.value_counts().sort_index()
    pct = (vc / vc.sum() * 100).round(2)
    return pd.DataFrame({"count": vc, "percent": pct})

print("Train severity distribution:")
display(dist(y_train))
print("Test severity distribution:")
display(dist(y_test))

# Numeric: median impute + scale
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler(with_mean=False))  # with_mean=False keeps sparse-friendly path
])

# Categorical: most-frequent impute + OHE with min_frequency to cap rare levels
ohe_kwargs = dict(handle_unknown="ignore")
try:
    # sklearn >= 1.1 supports min_frequency; safe performance guard for high-card columns
    ver = tuple(map(int, sklearn.__version__.split(".")[:2]))
    if ver >= (1, 1):
        ohe_kwargs["min_frequency"] = 0.01  # group categories <1% into 'other'
except Exception:
    pass

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(**ohe_kwargs))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
    ],
    remainder="drop"
)

# Fit on TRAIN only; transform both
X_train_enc = preprocessor.fit_transform(X_train)
X_test_enc  = preprocessor.transform(X_test)

print("\n Preprocessor fitted (train-only).")
print("Encoded train shape:", X_train_enc.shape)
print("Encoded test shape :", X_test_enc.shape)
print("Matrix type       :", type(X_train_enc).__name__)

# Report OHE feature expansion
ohe = preprocessor.named_transformers_["cat"].named_steps["ohe"]
ohe_out_dims = sum(len(cats) for cats in ohe.categories_)
print("Total OHE output features:", ohe_out_dims)

# Quick numeric NaN check post-imputation
num_pipe = preprocessor.named_transformers_["num"]
X_train_num_only = num_pipe.transform(X_train[numeric_features])
arr = X_train_num_only if isinstance(X_train_num_only, np.ndarray) else X_train_num_only.toarray()
print("Numeric NaNs after imputation:", int(np.isnan(arr).sum()))
```

```python
import time, numpy as np, pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, cohen_kappa_score,
    confusion_matrix, classification_report
)

# --- helpers ---
def evaluate(y_true, y_pred, label):
    acc   = accuracy_score(y_true, y_pred)
    f1_ma = f1_score(y_true, y_pred, average="macro")
    f1_wt = f1_score(y_true, y_pred, average="weighted")
    kappa = cohen_kappa_score(y_true, y_pred)
    return {"model": label, "accuracy": acc, "macro_f1": f1_ma, "weighted_f1": f1_wt, "kappa": kappa}

def print_cm_and_report(y_true, y_pred, labels=[1,2,3,4]):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    print("Confusion Matrix (rows=true, cols=pred):\n", cm)
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, digits=3))

results = []

# Dense copies for models that need dense input
X_train_dense = X_train_enc.toarray() if hasattr(X_train_enc, "toarray") else X_train_enc
X_test_dense  = X_test_enc.toarray()  if hasattr(X_test_enc, "toarray")  else X_test_enc
```

```python
# ---------------------------
# 1) Logistic Regression
# ---------------------------
t0 = time.time()
logit = LogisticRegression(
    multi_class="multinomial",
    solver="saga",
    class_weight=class_weight_map,   # from Step 4
    max_iter=500,
    n_jobs=-1
)
logit.fit(X_train_enc, y_train)
y_pred_logit = logit.predict(X_test_enc)
res = evaluate(y_test, y_pred_logit, "LogisticRegression")
res["time_s"] = time.time() - t0
results.append(res)

print("✅ LogisticRegression")
print(res)
print_cm_and_report(y_test, y_pred_logit)
```

```python
# ---------------------------
# 2) Random Forest — all defaults (n_estimators=100, etc.)
# ---------------------------

t0 = time.time()
rf = RandomForestClassifier()  # vanilla
rf.fit(X_train_dense, y_train)
y_pred_rf = rf.predict(X_test_dense)
res = evaluate(y_test, y_pred_rf, "RandomForest (vanilla)")
res["time_s"] = time.time() - t0
results.append(res)

print("\n✅ RandomForest (vanilla)")
print(res)
print_cm_and_report(y_test, y_pred_rf)
```
